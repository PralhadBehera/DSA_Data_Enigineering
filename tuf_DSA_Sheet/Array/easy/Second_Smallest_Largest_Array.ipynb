Find Second Smallest and Second Largest Element in an array


Problem Statement: Given an array, find the second smallest and second largest element in the array. Print ‘-1’ in the event that either of them doesn’t exist.

```python
import time

print('brute force approach')

start = time.time()  # start time

# Brute Force Approach
def find_second_smallest_largest_bruteforce(arr):
    if len(arr) < 2:
        return None, None

    # Remove duplicates and sort
    unique_arr = sorted(set(arr))

    if len(unique_arr) < 2:
        return None, None

    second_smallest = unique_arr[1]
    second_largest = unique_arr[-2]

    return second_smallest, second_largest

end = time.time()     # end time

print("Execution time:", end - start, "seconds")
```

⏱ Complexity

Time Complexity: O(n log n) (due to sorting)

Space Complexity: O(n) (extra space for set)

✅ Advantages

Very easy to understand

Beginner-friendly

❌ Disadvantages

Sorting is unnecessary when only second min/max is needed

```python
print('better approach')

start = time.time()  # start time
# Function to find the second smallest and second largest elements in the array
# Better Approach (Two Pass)
def find_second_smallest_largest_better(arr):
    n = len(arr)
    if n < 2:
        return None, None

    first_smallest = float('inf')
    first_largest = float('-inf')

    # First pass: find smallest and largest
    for num in arr:
        if num < first_smallest:
            first_smallest = num
        if num > first_largest:
            first_largest = num

    second_smallest = float('inf')
    second_largest = float('-inf')

    # Second pass: find second smallest and second largest
    for num in arr:
        if first_smallest < num < second_smallest:
            second_smallest = num
        if first_largest > num > second_largest:
            second_largest = num

    if second_smallest == float('inf'):
        second_smallest = None
    if second_largest == float('-inf'):
        second_largest = None

    return second_smallest, second_largest


        
    end = time.time()     # end time

print(find_second_smallest_largest_better([4, 2, 5, 1, 3]))

print("Execution time:", end - start, "seconds")
```

⏱ Complexity

Time Complexity: O(n)

Space Complexity: O(1)

✅ Advantages

No sorting

Efficient

Handles duplicates correctly

❌ Disadvantages

Requires two loops

```python
print('optimal Approach')

start = time.time()  # start time

# Optimal Approach (Single Pass)
def find_second_smallest_largest_optimal(arr):
    if len(arr) < 2:
        return None, None

    first_smallest = second_smallest = float('inf')
    first_largest = second_largest = float('-inf')

    for num in arr:
        # Smallest and second smallest
        if num < first_smallest:
            second_smallest = first_smallest
            first_smallest = num
        elif first_smallest < num < second_smallest:
            second_smallest = num

        # Largest and second largest
        if num > first_largest:
            second_largest = first_largest
            first_largest = num
        elif first_largest > num > second_largest:
            second_largest = num

    if second_smallest == float('inf'):
        second_smallest = None
    if second_largest == float('-inf'):
        second_largest = None

    return second_smallest, second_largest

end = time.time()     # end time

print(find_second_smallest_largest_optimal([4, 2, 5, 1, 3]))

print("Execution time:", end - start, "seconds")
```

⏱ Complexity

Time Complexity: O(n)

Space Complexity: O(1)

✅ Advantages

Fastest solution

Single loop

No extra space

Ideal for interviews

❌ Disadvantages

Slightly harder to understand for beginners
