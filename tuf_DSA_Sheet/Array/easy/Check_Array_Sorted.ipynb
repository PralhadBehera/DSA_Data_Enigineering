# Check if an Array is Sorted

## Problem Statement

Given an array of size **N**, write a program to check whether the array is sorted in  
**ascending / increasing / non-decreasing** order.

- Return **True** if the array is sorted  
- Return **False** otherwise  

An array is considered sorted if every element is **less than or equal to the next element**.

---

## Examples

### Example 1

Input:
N = 5  
array[] = {1, 2, 3, 4, 5}

Output:
True

Explanation:  
The given array is sorted. Every element in the array is smaller than or equal to its next value.

---

### Example 2

Input:
N = 5  
array[] = {5, 4, 6, 7, 8}

Output:
False

Explanation:  
The given array is not sorted because 5 is greater than 4, violating the non-decreasing order.


```python
print('brute Force')

def chck_sorted_brute(arr):
    n =len(arr)
    for i in range(n-1):
        if arr[i]>arr[i+1]:
            return False
    return True

print(chck_sorted_brute([1,2,3,4,5]))
```

## ✔ Methodology for Brute Force

- Use two nested loops
- For each element, compare it with all elements to its right
- If array[i] > array[j] for any j > i, return False

---

## ❌ Drawback forBrute Force

- Performs unnecessary comparisons
- Inefficient for large arrays


```python
print('\nBetter Approach')

def chck_sorted_better(arr):
    n = len(arr)
    if n == 0 or n == 1:
        return True
    for i in range(n-1):
        if arr[i] > arr[i + 1]:
            return False
    return True

print(chck_sorted_better([1,2,3,4,5]))
```

## ✔ Methodology Better Approach

- Traverse the array once
- Compare array[i] with array[i + 1]
- If any element is greater than the next one → not sorted

---

## ✅ Advantage Better Approach

- Much fewer comparisons than brute force
- Easy to understand

---

## ⚠ Limitation Better Approach

- Still checks even after sorting could already be determined early
