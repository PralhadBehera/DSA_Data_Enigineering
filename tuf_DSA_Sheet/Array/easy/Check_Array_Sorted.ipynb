# Check if an Array is Sorted

## Problem Statement

Given an array of size **N**, write a program to check whether the array is sorted in  
**ascending / increasing / non-decreasing** order.

- Return **True** if the array is sorted  
- Return **False** otherwise  

An array is considered sorted if every element is **less than or equal to the next element**.

---

## Examples

### Example 1

Input:
N = 5  
array[] = {1, 2, 3, 4, 5}

Output:
True

Explanation:  
The given array is sorted. Every element in the array is smaller than or equal to its next value.

---

### Example 2

Input:
N = 5  
array[] = {5, 4, 6, 7, 8}

Output:
False

Explanation:  
The given array is not sorted because 5 is greater than 4, violating the non-decreasing order.
