Find the Largest element in an array

```python
def largest_element(arr):
    n = len(arr)
    max_value = arr[0]
    for i in range(1 , n):
        if arr[i]> max_value:
            max_value = arr[i]
    return max_value
```

```python
print(largest_element([1, 2, 3, 0, -1, 5, 4]))  # Output: 5
```
